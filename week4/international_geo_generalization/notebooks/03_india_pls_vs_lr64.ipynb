## Takeaways

1. **Use PLS-2 as the India transfer baseline**, not raw lr64 — mean LORO 0.60 vs 0.52.
2. **Gap to USA remains large** (0.60 vs ~0.85) — low-rank supervision mitigates fragmentation but does not create USA-like sharing.
3. **PLS-9 ≈ middle ground** — better than lr64 on LORO, worse than PLS-2 (extra components reintroduce city-specific noise).
