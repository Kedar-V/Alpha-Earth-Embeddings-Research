## Takeaways

1. **Japan 10-city LORO mean lr64 F1 = 0.770** — similar to prior 5-city (0.730), between USA (0.860) and India (0.523).
2. **Sharing, not fragmentation:** LORO ≈ within (0.796); city→city off-diag 0.730 (India was 0.45).
3. **Sapporo is the weak pole** (0.585); Yokohama strongest (0.846).
4. **PLS ≈ lr64** for Japan — use lr64 as the Japan baseline; reserve PLS for India-style fragmented panels.
