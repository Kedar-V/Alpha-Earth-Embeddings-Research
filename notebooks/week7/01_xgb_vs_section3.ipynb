# Week 7 results — XGB vs Section 3 (Ridge eval sets)

Loads `outputs/week7_section3_xgb_mae/latest_run/` when present.


```
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display, Markdown

ROOT = Path('../..').resolve()
run = ROOT / 'outputs' / 'week7_section3_xgb_mae' / 'latest_run'
if not run.exists():
    # fallback newest
    cands = sorted((ROOT/'outputs'/'week7_section3_xgb_mae').glob('20*'))
    run = cands[-1] if cands else None
print('run:', run)

if run and (run/'summary.json').exists():
    s = json.loads((run/'summary.json').read_text())
    print('protocol:', s.get('protocol'))
    print('buildings n:', s['building']['n'])
    print('tiles n:', s['tiles_100m']['n'])
    b = pd.read_csv(run/'artifacts'/'mae_buildings.csv')
    t = pd.read_csv(run/'artifacts'/'mae_tiles_100m.csv')
    display(Markdown('### Buildings (soft)'))
    display(b[b.gt=='soft'][['model','n','res','nres','mean']])
    display(Markdown('### 100 m tiles (soft)'))
    display(t[['model','n','res','nres','mean']])
    fig = run/'figures'/'section3_mae_headline.png'
    if fig.exists():
        display(Image(filename=str(fig)))
else:
    print('No completed week7 Section3 run yet — submit after XGB train finishes.')
```
