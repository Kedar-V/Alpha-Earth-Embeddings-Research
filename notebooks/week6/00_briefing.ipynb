Fair comparison of **AlphaEarth fraction heads** vs **GHS-BUILT-S** using **Overture building polygons** as the common spatial support.

Motivation: sampling GHSL at AE cell centers forces GHSL onto AE's lattice (~4 m origin shift). Here both products are summarized independently into each Overture footprint — AE via per-metro UTM cells, GHSL via Mollweide — then scored with MAE on `(frac_res, frac_nres)`.

| Piece | Choice |
|---|---|
| Reference | Overture Track-A footprints (`resi_def_v1`) |
| GT | `y_res` → `(1,0)` / `(0,1)` |
| AE | Mean week4 Ridge LORO OOF on `inside_polygon` cells |
| GHSL | GHS-BUILT-S soft fractions under polygon mask |
| Metric | `mae_mean_res_nres` on shared coverage |

Results notebook: [`01_ae_vs_ghsl_overture_polygons.ipynb`](01_ae_vs_ghsl_overture_polygons.ipynb).
